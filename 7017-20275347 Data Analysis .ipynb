```python
#Creating a cognative Matrix 
import pandas as pd
from datetime import datetime

# proper datetime format
data['event_time'] = pd.to_datetime(data['event_time'])

# Calculating Duration (D)
duration = data.groupby('user_id')['event_time'].agg(lambda x: (x.max() - x.min()).total_seconds() / 60)  # Duration in minutes

# Calculating Visiting Frequency (VF)
visiting_frequency = data.groupby('user_id')['user_session'].nunique()

# Function to calculate HHI index
def calculate_hhi(series):
    total = series.sum()
    proportions = series / total
    hhi = (proportions ** 2).sum()
    return hhi

# Calculating Product Category Consistency (PCC)
category_counts = data.groupby(['user_id', 'category_id']).size()
pcc = category_counts.groupby(level=0).apply(calculate_hhi)

# Calculating Brand Consistency (BC)
brand_counts = data.groupby(['user_id', 'brand']).size()
bc = brand_counts.groupby(level=0).apply(calculate_hhi)

# Combine all metrics into a single DataFrame
results = pd.DataFrame({
    'Duration': duration,
    'Visiting Frequency': visiting_frequency,
    'Product Category Consistency': pcc,
    'Brand Consistency': bc
}).reset_index()

# Output the results to a CSV file
results.to_csv('user_cognitive_metrics.csv', index=False)
print("Metrics calculated and saved to user_cognitive_metrics.csv")
```

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from scipy.stats import norm

data = pd.DataFrame({
    'Duration': np.random.normal(loc=60, scale=10, size=1000),
    'Visiting Frequency': np.random.normal(loc=120, scale=20, size=1000),
    'Product Category Consistency': np.random.normal(loc=30, scale=5, size=1000),
    'Brand Consistency': np.random.normal(loc=50, scale=5, size=1000)
})

# Function to plot the normal distribution
def plot_normal_dist(data, column_name):
    plt.figure(figsize=(10, 6))
    sns.distplot(data[column_name], fit=norm, kde=True)
    plt.title('Normal distribution of ' + column_name)
    plt.xlabel(column_name)
    plt.ylabel('Frequency')
    plt.show()

# Plotting the normal distribution for each column
for column in data.columns:
    plot_normal_dist(data, column)
```

```python
import matplotlib.pyplot as plt


df = pd.read_csv('user_cognitive_metrics.csv')

# Calculating the mean, median, and 75th percentile
mean_duration = df['Duration'].mean()
median_duration = df['Duration'].median()
percentile_75th = df['Duration'].quantile(0.75)

# Creating the histogram
plt.figure(figsize=[10,8])
df['Duration'].hist(bins='auto', color='#0504aa', alpha=0.7, density=True)

# Adding lines for the mean, median, and 75th percentile
plt.axvline(mean_duration, color='r', linestyle='dashed', linewidth=2, label=f'Mean: {mean_duration:.2f}')
plt.axvline(median_duration, color='g', linestyle='dashed', linewidth=2, label=f'Median: {median_duration:.2f}')
plt.axvline(percentile_75th, color='b', linestyle='dashed', linewidth=2, label=f'75th percentile: {percentile_75th:.2f}')

# Adding legend, title and labels
plt.legend()
plt.title('Histogram of Duration')
plt.xlabel('Duration')
plt.ylabel('Density')
plt.grid(axis='y', alpha=0.75)
plt.show()
```
